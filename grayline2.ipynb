## Cleaning the data

```python
# Clean the DriverName column
final_df['DriverName'] = (
    final_df['DriverName']
    .str.replace(',', '', regex=False)         # Remove commas
    .str.strip()                               # Remove leading/trailing whitespace
    .str.replace(r'\s+', ' ', regex=True)      # Collapse multiple spaces into one
)

# Converting date time to date only
final_df['AccidentDate'] = pd.to_datetime(final_df['AccidentDate'], errors='coerce')

#Fixing the multiple column values to 'Y' and 'N' for 'DOT Recordable'
final_df['DOT Recordable'] = final_df['DOT Recordable'].replace({'Non-DOT':'N', 
                                                                 'DOT':'Y', 
                                                                 'FALSE':'N',
                                                                 'TRUE':'Y',
                                                                 False:'N', 
                                                                 True: 'Y', 
                                                                 'non-DOT':'N', 
                                                                 'NON-DOT':'N', 
                                                                 'NON DOT': 'N', 
                                                                 '?':'Not Registered',
                                                                 'NON -DOT':'N'})

final_df['DOT Recordable'] = final_df['DOT Recordable'].fillna('Not Registered')

#Fixing the multiple column values of 'IncidentAccident' column
final_df['IncidentAccident'] = final_df['IncidentAccident'].replace({'?':'Not Registered',
                                                                     'Accicdent':'Accident',
                                                                     'Accidednt':'Accident',
                                                                     'Accident?':'Accident?',
                                                                     'Accidnet':'Accident',
                                                                     'INCIDENT':'Incident',
                                                                     'ACCIDENT':'Accident',
                                                                     'Incident / Accident':'Incident / Accident',
                                                                     'Mystery':'Mystery'})
final_df['IncidentAccident'] = final_df['IncidentAccident'].fillna('Not Registered')

#Fixing the multiple column values of 'PoliceReport' column
final_df['PoliceReport'] = final_df['PoliceReport'].replace({'No':'N',
                                                             'Yes':'Y',
                                                             'FALSE':'N',
                                                             'TRUE':'Y',
                                                             'Na':'Not Registered'})

##Fixing the multiple column values of 'SafetyCommRuling' column

final_df['Prevent'] = final_df['SafetyCommRuling'].combine_first(final_df['Preventable'])
final_df['Prevent'] = final_df['Prevent'].replace({'Non-Preventable':'N',
                                                   'Preventable':'Y',
                                                   'o-Prevetable':'N',
                                                   'Non-Preventable ':'N',
                                                   'no':'N',
                                                   'yes':'Y',
                                                   'Yes':'Y',
                                                   'True':'Y',
                                                   'No':'N',
                                                   'False':'N'})

final_df = final_df.drop(columns=['SafetyCommRuling', 'Preventable'])
                                                                     





                                                               



```
