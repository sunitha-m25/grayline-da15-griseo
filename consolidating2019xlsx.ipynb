```python
consolidatedaccidents2019_df = consolidatedaccidents2019_df.groupby('ClaimNo', as_index=False).agg(lambda x: ' | '.join(map(str, x)))
```

```python
def extract_first_valid_value(cell):
    if pd.isna(cell):
        return np.nan
    parts = [p.strip() for p in str(cell).split('|')]
    for part in parts:
        if part.lower() != 'nan' and part != '':
            return part
    return np.nan

# Apply to entire DataFrame
final2019_df = consolidatedaccidents2019_df.applymap(extract_first_valid_value)

# Optional: Drop fully empty columns
final2019_df = final2019_df.dropna(axis=1, how='all')

# Preview result
final2019_df.head()    
```
