```python
# Creating a unique ID for drivername
def assign_sequential_ids(DriverName):
    assigned_ids = {}
    counter = 1
    for entry in DriverName:
        if entry not in assigned_ids:
            assigned_ids[entry] = f"{counter:04d}"
            counter += 1
    return assigned_ids

DriverId = assign_sequential_ids(final_df['DriverName'])

final_df['DriverId'] = final_df['DriverName'].map(DriverId)
print(DriverId)
```
